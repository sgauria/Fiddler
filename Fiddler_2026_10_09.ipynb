### This Week’s Fiddler
A surfer is riding along a crest of a sinusoidal wave given by the function y = 2·sin(x−t), where y is the height above sea level in meters, x is horizontal position in meters, and t is time in seconds. Based on this, the surfer knows that her speed—and the speed of the wave—is 1 meter per second.

But hold on—it appears that another wave is approaching from the opposite direction. This wave has half the amplitude but twice the speed: y = sin(x+2t).

The two waves, the forward and the backward, ”interfere,” meaning they combine. In other words, the height of the water at position x and time t is given by y = 2·sin(x−t) + sin(x+2t). Nevertheless, the surfer stays atop a crest of the combined wave.

As the wave moves along, what are the surfer’s slowest and fastest speeds, in meters per second? (Edit: My intent was to ask for speed in the horizontal direction. If you include horizontal and vertical components, I’ll still mark it as correct.)

# Solution

Let's call the (height of the) wave W.

$ W(x,t) = 2 sin(x-t) + sin(x+2t) $

The maxima (crests) of the wave are points where W' = 0 and W'' is negative.

$ W' = \frac{dW}{dx} = 2 cos(x-t) + cos(x+2t) $

$ W'' = \frac{dW'}{dx} = -2 sin(x-t) -sin(x+2t) $

$a + b cos(\phi) = C cos(\theta)$

So:

$ tan(\theta) = (b sin(\phi))/(a + b cos(\phi)) $

$ C^2 = a^2 + 2ab*cos(\phi) + b^2$

#### Translating back

In the original puzzle: 

$ a = 2, b = 1, u = x-t, \phi = 3t$

So:

$tan(\theta) = sin(3t)/(2+cos(3t))$

i.e.

$\theta = tan^{-1}(sin(3t)/(2+cos(3t)))$

And:

$ C^2 = 5 + 4 cos(3t) $

i.e. 

$ C = \sqrt{5 + 4 cos(3t)} $

In other words:

$ W(x,t) = 2 sin(x-t) + sin(x+2t) = \sqrt{5 + 4 cos(3t)} sin( x - t + tan^{-1}(sin(3t)/(2+cos(3t))) ) $

It's not a whole lot simpler than earlier attempts, but the structure is better suited for our needs.


#### From waveform to crest position

With this structure, we can see that, at any time t, the height of the crest is $\sqrt{5 + 4 cos(3t)}$ and the position of the crests is at $x - t + tan^{-1}(sin(3t)/(2+cos(3t))) = \pi/2 + 2n\pi$.

So, the surfer's (x,y) position:

$ S(t) = ( \pi/2 + t - tan^{-1}(sin(3t)/(2+cos(3t)))  , \sqrt{5 + 4 cos(3t)}  ) $

The surfer's velocity V would be the derivative wrt t.

$V(t) = \frac{dS}{dt} $

These min/max values are the same whether you consider just velocity along the x direction, or the x and y combined velocity.


Average speed along the x direction is actually easier to calculate than the min/max.

Since all the trignometric expressions in $S(t)$ are functions of $3t$, everything repeats every $2\pi/3$ seconds. 

So, if we calculate $S(t + 2\pi/3) - S(t)$, all the constant and trignometric terms will cancel. Only the t term won't cancel. So, the displacement in this time will be just $(t + 2\pi/3) - (t) = 2\pi/3$, i.e. equal to the time interval. 

So, the average speed is just 1 meter / sec.
