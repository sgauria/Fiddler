### This Week’s Fiddler
Inspired by Delta Airlines, you and a graphic designer colleague are crafting a new logo for Fiddler Airlines. Your colleague starts with an equilateral triangle with a side length of 1 unit. Next, you pick a random point inside the triangle, giving you four points: the three corners of the triangle and this fourth point inside. There are several (simple) quadrilaterals that can be drawn whose four vertices are these four points. Your colleague picks one of these quadrilaterals at random, with each having an equal probability. This quadrilateral will be the new logo.

On average, what can you expect the area of the quadrilateral to be?

### This Week’s Extra Credit
While picking random points inside the equilateral triangle with side length 1, you and your colleague notice that some points are “special.” When one of these special points is connected with the triangle’s three vertices, it creates three smaller triangles whose areas form an arithmetic sequence.

After further investigation, you realize these special points all belong to a locus, the total length of which is L. What is this length L?

# Fiddler Solution

So, we are picking a point at random. In picking a quadrilateral, we are effectively picking an edge of the original triangle and removing the triangle formed by that edge and the new point. So, we can effectively pick the edge first, call it the base, and then pick a point and remove the triangular area between the base and the chosen point.

The area of the new triangle is a linear function of the height of the point from the base. So, the average area removed will be the area of the triangle formed using the average point, which is the centroid. The height of the centroid is 1/3 the height of the whole triangle. So, the average area removed is 1/3 of the original area, and the average area of the quadrilateral is 2/3 of the original area. i.e. 

$ E(A_{quad}) = 2/3 * (1/2 * 1 * \sqrt{3} / 2) $

$ E(A_{quad}) = \sqrt{3} / 6 = 0.28867513459481288225457439025098 $

# Fiddler Solution Alt

That was a so quick and slick, that I want to cross-check it. Let's put our origin at the center of the base.

$A_{eq_tri} = \sqrt{3}/4$

$A_{new_tri} = y/2$

$A_{quad} = \sqrt{3}/4 - y/2$

At a point y, the x range is from $-(1/2-y/\sqrt{3})$ to $+(1/2-y/\sqrt{3})$

Now the integral of that is:

$I = \int^{\sqrt{3}/2}_0 dy \int_{-(1/2-y/\sqrt{3})}^{+(1/2-y/\sqrt{3})} dx A_{quad} $

$I = \int^{\sqrt{3}/2}_0 dy (1-2y/\sqrt{3}) (\sqrt{3}/4 - y/2) $

$I = \int^{\sqrt{3}/2}_0 dy (\sqrt{3}/4  -y/2 -y/2  + y^2/\sqrt{3}) $

$I = \int^{\sqrt{3}/2}_0 dy (\sqrt{3}/4  -y  + y^2/\sqrt{3}) $

$I = (\sqrt{3}/4y  -y^/2  + y^3/3\sqrt{3}) |^{\sqrt{3}/2}_0 $

$I = (3/8  - 3/8  + 3\sqrt{3}/(8*3*\sqrt{3}) $

$I = 1/8$

And the normalizing factor is

$N = \int^{\sqrt{3}/2}_0 dy \int_{-(1/2-y/\sqrt{3})}^{+(1/2-y/\sqrt{3})} dx 1 $

$N = \int^{\sqrt{3}/2}_0 dy (1-2y/\sqrt{3}) $

$N = (y  - y^2/\sqrt{3}) |^{\sqrt{3}/2}_0 $

$N = (\sqrt{3}/2 - 3/(4*\sqrt{3}))$ 

$N = (\sqrt{3}/2 - \sqrt{3}/4) $

$N = \sqrt{3}/4 $

So, the expected value:

$E(A_{quad}) = I/N = 1/(2*\sqrt{3}) = \sqrt{3}/6$

Okay, it's consistent. Which could be consistently wrong, but it is consistent. :)

# Extra Credit Solution.

We choose a random point P inside triangle ABC. 

This divdes the triangle into triangles PBC, PCA, PAB.

Let the 3 altitudes from P to BC,CA,AB be $h_a, h_b, h_c$.

Since PAB, PBC, PCA divides the original triangle into 3 parts, the sum of their areas should equal the area of ABC.i.e.

$1/2 * 1 *  (h_a + h_b + h_c) = 1/2 * 1 * \sqrt{3}/2 $

$h_a + h_b + h_c = \sqrt{3}/2 $

Now, if the areas form an arithmetic sequence, the altitudes $h_a, h_b, h_c$ must also form an arithmetic sequence.

So, let the altitudes be s-d, s, s+d in some order. So, we can also see that.

$h_a + h_b + h_c = (s-d)+s+(s+d) = 3s $

Putting that together, we have

$3s = \sqrt{3}/2 $

$s = \sqrt{3}/6 $

So, if one of the altitudes is this special value s, then the altitudes will form an arithmetic sequence, and so will the areas.

This value s is also the height of the centroid above the base.

So, the locus consists of 3 lines parallel to and at height s from each of the sides.

To calculate the length of each line, we recall this fact from the alt solution above: *At a point y, the x range is from $-(1/2-y/\sqrt{3})$ to $+(1/2-y/\sqrt{3})$*

So, the length of the line is $L_{line} = (1-2y/\sqrt{3})$. Plugging in $y=s=\sqrt{3}/6 $, we get $L_{line} = (1-(2/6)) = (1-(1/3)) = 2/3$. 

That makes sense, we are 2/3 of the distance from the top, so the length is the base reduced by a factor 2/3.

So, the length of the entire locus is 2/3*3 = 2.
