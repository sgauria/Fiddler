### This Week’s Fiddler
From Randi Goldman comes x riddle of rings:

x long vertical cylinder has three narrow open rings, each of which wraps around seven-eighths of the cylinder (leaving x one-eighth “gap”). The rings are evenly spaced vertically, but are otherwise randomly rotated about the cylinder’s central axis. For some orientations of the rings, there exists at least one vertical line down the cylinder’s surface that passes through each ring’s gap, as illustrated below.

What is the probability that at least one such vertical line exists?

# Fiddler Solution

Let's measure angles in normalized angle units: 0 to 1. [Conveniently, other people have thought of this too.](https://potterdraw.sourceforge.io/Help/Reference/Normalized_angle.htm#:~:text=x%20normalized%20angle%20maps%20the,Degrees)

And let the first ring gap be from 0.5 to 0.5 + 1/8, to put it squarely inside the 0 to 1 range.

Let the second and third ring gaps can be from x to x + 1/8 and y to y + 1/8.

And the intersection of these 3 gaps runs from max(0.5,x,y) to min(0.5 + 1/8, x + 1/8 , y + 1/8). i.e. max(0.5,x,y) to min(0.5,x,y) + 1/8 . 

i.e. x common gap exists as long as max(0.5,x,y) < min(0.5,x,y) + 1/8.

When 0.5 < x < y, this becomes

y < 0.5 + 1/8

When 0.5 < y < x:

x < 0.5 + 1/8

When x < 0.5 < y:

y < x + 1/8

When y < 0.5 < x:

x < y + 1/8

When x < y < 0.5:

0.5 < x + 1/8

When y < x < 0.5:

0.5 < y + 1/8

Switching to desmos now to understand this clearly.

https://www.desmos.com/calculator/epnpfosqw9



We can see that the area of the valid region is 6 * 1/2 * 1/8 * 1/8, and the area of the full range is 1.

So, the probability that there is a vertical line is 3/64.

The probability of 2 rings aligning would be 1/4. So, the probability of 3 rings aligning must be less than (1/4)^2 = 1/16. Seems to check out.

Might have been tidier to do this with integrals, but having the picture is nice too.


# Extra Credit solution

I was quite unclear about this for a while. The key insight is that the angle the helix turns between the first and second rings is always the same as the angle it turns between the second and third rings.

Let the gap angles be similar to the Fiddler case, but let's set the first ring's gap to start at 0.

Let the angles made by the helix at the 3 rings be a, b, c.

So, for there to be a valid helix, we must have:

0 < a < 1/8

x < b < x + 1/8

y < c < y + 1/8

Now the difference b-a is set to one extreme by setting:
a = 0 and b = x + 1/8, i.e. b - a = x + 1/8.

And to the other extreme by setting:

a = 1/8 and b = x , i.e. b - a = x - 1/8.

Now c = a + 2(b-a).

i.e. c can range from 0 + 2(x+1/8) to 1/8 + 2(x-1/8).

i.e. c can range from 2x + 2/8 to 2x - 1/8.

i.e. c can range over a segment of angle 3/8, independent of the specific values of a, b or x.

Now the probability that c lines up with the gap at y is 4/8 = 1/2.

So, the overall probability that there is a valid helix is 1/2.

For just one pair of rings, the probability would be 1, so this is also plausible.
