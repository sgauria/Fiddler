# Fiddler Solution

It makes sense to divide this into cases.

#### Diagonal case
The 2 points are sure to be labelled differently here.

Top point can be labelled 8 ways, bottom point can be labelled 3 ways.

The connector can be chosen 2 ways.

Each case in the top-left + bottom-right case is a reflection of a case in the top-right + bottom-left.

So, the number of distinct diagrams are 8 * 3 * 2 = 48.

#### Vertical case.

Pretty much identical to the diagonal case. 

So a further 48 distinct diagrams here.

#### Horizontal top case

The 2 points can be labelled identically 8 * 1 = 8 ways, and labelled distinctly 8 * 7 = 56 ways.

In the identical case, we must use the green connector.

In the distinct case, we can use either the green or the yellow connector.

But in the distinct case, pairs are reflections of each other (e.g. A-yellow-B = B-yellow-A)

So, the number of distinct diagrams is 8 + (56 * 2 / 2) = 8 + 56 = 64.

#### Horizontal bottom case

The 2 points can be labelled identically 3 * 1 = 3 ways, and labelled distinctly 3 * 2 = 6 ways.

In the identical case, we can use the green or blue connector.

In the distinct case, we can use either the green or the yellow connector.

In the distinct case, pairs are reflections of each other (e.g. J-yellow-K = K-yellow-J)

So, the number of distinct diagrams is (3 * 2) + (6 * 2 / 2) = 6 + 6 = 12.

#### All together now

Total number of distinct diagrams is 48 + 48 + 64 + 12 = **172**.

So, I think we need to count the number of distinct diagram possible (now without reflection and with more options), keeping track of the connector colors and the possibilities that they could match some diagram from the fiddler list (now augment with random secret letters). Let's give it a shot.

Letters can be chosen 9 * 9 * 4 * 4 = 1296 ways, secrecy can be chosen 6 ways, and then the connector can be chosen 4 ways. So, there are 1296 * 24 = 31104 possible diagrams. 

But this doesn't address the colors properly. So let's try to divide this by color.

#### Gray = 0 pts.

Gray is valid for all the links, but the top link has gray twice. For simplicity, let's call one of those links gray2.

Letters can be chosen 1296 ways and secrecy can be chosen 6 ways, and the connector must be gray.

So, there are 7776 diagrams with a gray connector.

In addition, there are 9 * 9 * 4 * 4 = 1296 diagrams with a gray2 connector.

So, 9072 diagrams in all.

In the Fiddler list, there were 0 diagrams with a gray connector.

#### Yellow = 1 pt.

Yellow is also valid for all the links.

Letters can be chosen 1296 ways and secrecy can be chosen 6 ways, and the connector must be yellow.

So, there are 7776 diagrams with a yellow connector.

In the fiddler list, there were 79 yellow diagrams.

#### Red = $\sqrt{2}$ pts.

Red is valid for vertical and diagonal links.

So, there must be exactly one visible top point and one visible bottom point, i.e. secrecy can be chosen 4 ways.

So, there 1296 * 4 = 5184 diagrams with a red connector.

In the fiddler list, there were 48 red diagrams.

#### Green = $\sqrt{3}$ pts.

Green is valid for horizontal links only

So, there must be either both top pts are visible, or both bottom points, i.e. secrecy can be chosen 2 ways.

So, there 1296 * 2 = 2592 diagrams with a green connector.

In the fiddler list, there were 42 green diagrams.

#### Blue = $\sqrt{6}$ or 0 pts.

Blue is valid for vertical, diagonal and bottom horizontal links, with pts only for identical pairs.

For the vertical and diagonal links, the links always connect different letters.

For the bottom horizontal link, it can connect either similar or different letters.

Similar to red, there are 5184 diagrams with a vertical or diagonal blue connector.

Analogous to green, there are 1296 diagrams with a bottom blue connector of which 81 * 12 = 972 are connecting different letters and 81 * 4 = 324 are connecting identical letters.

So, there are 5184 + 972 = 6156 blue diagrams that are worth 0 pts.

And there are 324 blue diagrams worth $\sqrt{6}$ pts.

In the fiddler list, there were 3 blue diagrams, all worth $\sqrt{6}$ pts.

#### Double check

9072   +   7776   +   5184   +   2592   +   6156   +   324 == 31104.

So, we appear to have counted correctly.
