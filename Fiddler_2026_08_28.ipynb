### This Week’s Fiddler
In Spelling Bee, a word game from The New York Times, you must create words using seven letters arranged in a honeycomb grid. Letters can be used more than once per word, but each word must be at least four letters long and include the central letter. You may recall a previous puzzle (RIP, FiveThirtyEight) based on Spelling Bee.

Each day, your goal is to score as many points as possible. Four-letter words are worth 1 point, longer words are worth the number of letters they contain, and “pangrams” (words containing every letter) provide 7 bonus points. But these specific details don’t matter for this week’s puzzle.

If you find all the words in a given day and therefore accrue the maximum number of points, you earn the “Queen Bee” ranking. Meanwhile, accruing smaller point totals earns you other rankings. In particular, the point cutoff for “Genius” is 70 percent of the maximum number of points, rounded to the nearest whole number. While the cutoff for “Genius” is clearly displayed in the puzzle, the “Queen Bee” point total is not readily shown.

Of course, this maximum total can be approximated by dividing the “Genius” cutoff by 0.7. Even so, the total may be ambiguous, since multiple “Queen Bee” values can result in the same “Genius” cutoff.

Suppose a given round of Spelling Bee has some very large, randomly chosen point total. What is the probability that this total can be precisely determined (i.e., without any ambiguity) from its point cutoff for “Genius”?

# Solution

While the fiddler could be done by hand (and I think it is 6 ambigous to 4 unambiguous), the EC likely needs code for one to be sure, so I am just going the code route.

```python
def round_half_to_pinf(x):
    return int(x + 0.5)

def cutoff_from_fullscore(pct, fullscore):
    cutoff_unrounded = fullscore * pct / 100.0
    return round_half_to_pinf(cutoff_unrounded)

def get_cutoffs(pcts, fullscore):
    cutoffs = [cutoff_from_fullscore(pct, fullscore) for pct in pcts]
    return tuple(cutoffs)

def get_range_maps(pcts, min_score=100, max_score=200):
    full_to_cutoffs = {}
    cutoffs_to_full = {}
    for full in range(min_score, max_score):
        cutoffs = get_cutoffs(pcts, full)
        full_to_cutoffs[full] = cutoffs
        if cutoffs not in cutoffs_to_full:
            cutoffs_to_full[cutoffs] = []
        cutoffs_to_full[cutoffs].append(full)
    return full_to_cutoffs, cutoffs_to_full

import math
def one_recurrence_interval(p):
    i = 1
    while i * 100 % p != 0:
        i += 1
    return int((i * 100) / p)

def analyze_cutoffs(pcts, scale=10):
    recurrence_intervals = [one_recurrence_interval(p) for p in pcts]
    recurrence_interval = math.lcm(*recurrence_intervals)
    print(f"Combined Recurrence interval: {recurrence_interval}")
    min_score = recurrence_interval * 1 * scale
    min_counting_threshold = recurrence_interval * 2 * scale
    max_counting_threshold = recurrence_interval * 3 * scale
    max_score = recurrence_interval * 4 * scale
    full_to_cutoffs, cutoffs_to_full = get_range_maps(pcts, min_score, max_score)
    num_full_scores = 0
    num_unique_cutoffs = 0
    ambiguous_cutoffs = 0
    ambiguous_full_scores = 0
    unambiguous_cutoffs = 0
    unambiguous_full_scores = 0
    for cutoffs, full_scores in cutoffs_to_full.items():
        if min_counting_threshold <= full_scores[0] < max_counting_threshold:
            num_full_scores += len(full_scores)
            num_unique_cutoffs += 1
            len_full_scores = len(full_scores)
            #print(f"Cutoff: {cutoff}, Full scores: {full_scores}")
            if len_full_scores > 1:
                ambiguous_cutoffs += 1
                ambiguous_full_scores += len_full_scores
            else:
                unambiguous_cutoffs += 1
                unambiguous_full_scores += len_full_scores
    print(f"Full scores: Total: {num_full_scores}, Ambiguous: {ambiguous_full_scores}, Unambiguous: {unambiguous_full_scores}, Unambiguous probability: {1.0 * unambiguous_full_scores / num_full_scores}")
    print(f"    Cutoffs: Total: {num_unique_cutoffs}, Ambiguous: {ambiguous_cutoffs}, Unambiguous: {unambiguous_cutoffs}, Unambiguous probability: {1.0 * unambiguous_cutoffs / num_unique_cutoffs}")
```

```python
# Fiddler
for scale in [1, 2, 5, 10, 20, 50, 100]:
    analyze_cutoffs([70], scale=scale)
```

```python
# Extra credit
for scale in [1, 2, 5, 10, 20, 50, 100]:
    analyze_cutoffs([2,5,8,15, 25, 40, 50], scale=scale)
```
